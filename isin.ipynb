```python
import pandas as pd
```

```python
data = pd.read_csv("deliveries.csv")
```

```python
death_over = data[data["over"]>15]
all_players=death_over.groupby("batsman")["ball"].count()
above200 = all_players[all_players>200].index.tolist() #give the list of those players who has played more than 200 balls in over 16-20
```

```python
#isin function gives that dataframe which satisfy provided list conditions
total_runs = death_over[death_over["batsman"].isin(above200)].groupby("batsman")["batsman_runs"].sum()
total_balls = death_over[death_over["batsman"].isin(above200)].groupby("batsman")["ball"].count()
```

```python
strate = (total_runs/total_balls) * 100
```

```python
strate.sort_values(ascending=False)
```
