```python
import pandas as pd
```

```python
deliveries = pd.read_csv("deliveries.csv")
match = pd.read_csv("matches.csv")
```

```python
newData = deliveries.merge(match , left_on="match_id" , right_on="id") #Merge two Datasets
```

```python
#Code to find orange cap of each ipl seasons
eachSeasonBatsman = newData.groupby(["season" , "batsman"])["batsman_runs"].sum() #Make multi-indexed Series
topBatsman = eachSeasonBatsman.sort_values(ascending=False).reset_index() #converts the series to dataframe
topBatsman.drop_duplicates(subset="season" , keep="first").sort_values("season")[["season" , "batsman"]]
```
